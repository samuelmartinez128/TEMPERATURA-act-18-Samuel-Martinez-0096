## 1. Fórmulas de Conversión

Para convertir entre estas dos escalas, utilizamos las siguientes relaciones matemáticas:

* **De Celsius (°C) a Fahrenheit (°F):**
    $$°F = (°C \times \frac{9}{5}) + 32$$
    *(O también: $°F = (°C \times 1.8) + 32$)*

* **De Fahrenheit (°F) a Celsius (°C):**
    $$°C = \frac{°F - 32}{1.8}$$
    *(O también: $°C = (°F - 32) \times \frac{5}{9}$)*

---

## 2. Ejemplos Resueltos Paso a Paso

### Ejemplo 1: Convertir 25°C a °F (Temperatura ambiente)
1.  **Identificar la fórmula:** $°F = (°C \times 1.8) + 32$
2.  **Sustituir el valor:** $°F = (25 \times 1.8) + 32$
3.  **Multiplicar:** $25 \times 1.8 = 45$
4.  **Sumar 32:** $45 + 32 = 77$
* **Resultado:** 25°C equivalen a **77°F**.

### Ejemplo 2: Convertir 98.6°F a °C (Temperatura corporal promedio)
1.  **Identificar la fórmula:** $°C = (°F - 32) / 1.8$
2.  **Sustituir el valor:** $°C = (98.6 - 32) / 1.8$
3.  **Restar 32:** $98.6 - 32 = 66.6$
4.  **Dividir entre 1.8:** $66.6 / 1.8 = 37$
* **Resultado:** 98.6°F equivalen a **37°C**.

## 3. Código en Python

Este programa utiliza funciones específicas para cada cálculo y un bucle `while` para mantener el menú activo hasta que decidas salir.

```python
def convertirCF():
    celsius = float(input("\nIngrese los grados Celsius a convertir: "))
    fahrenheit = (celsius * 1.8) + 32
    print(f"--- {celsius}°C es igual a {fahrenheit:.2f}°F ---")

def convertirFC():
    fahrenheit = float(input("\nIngrese los grados Fahrenheit a convertir: "))
    celsius = (fahrenheit - 32) / 1.8
    print(f"--- {fahrenheit}°F es igual a {celsius:.2f}°C ---")

def menu():
    while True:
        print("\n--- MENÚ DE CONVERSIONES ---")
        print("1. Convertir grados Centígrados a Fahrenheit")
        print("2. Convertir grados Fahrenheit a Centígrados")
        print("3. Salir")
        
        opcion = input("Seleccione una opción (1-3): ")

        if opcion == '1':
            convertirCF()
        elif opcion == '2':
            convertirFC()
        elif opcion == '3':
            print("Saliendo del programa... ¡Hasta luego!")
            break
        else:
            print("Opción no válida. Intente de nuevo.")

# Llamada principal para iniciar el programa
if __name__ == "__main__":
    menu()
```

### Explicación breve del código:
* **`float(input(...))`**: Permite que el usuario ingrese números con decimales.
* **`f"{valor:.2f}"`**: Es un formato para que el resultado solo muestre **2 decimales**, lo cual es más limpio.
* **`while True`**: Crea un bucle infinito que solo se rompe cuando el usuario elige la opción 3 y el programa ejecuta el `break`.
* **Modularidad**: Al separar los cálculos en `convertirCF()` y `convertirFC()`, el código es más fácil de leer y mantener.

````python
def convertirCF():
    celsius = float(input("\nIngrese los grados Celsius a convertir: "))
    fahrenheit = (celsius * 1.8) + 32
    print(f"--- {celsius}°C es igual a {fahrenheit:.2f}°F ---")

def convertirFC():
    fahrenheit = float(input("\nIngrese los grados Fahrenheit a convertir: "))
    celsius = (fahrenheit - 32) / 1.8
    print(f"--- {fahrenheit}°F es igual a {celsius:.2f}°C ---")

def menu():
    while True:
        print("\n--- MENÚ DE CONVERSIONES ---")
        print("1. Convertir grados Centígrados a Fahrenheit")
        print("2. Convertir grados Fahrenheit a Centígrados")
        print("3. Salir")

        opcion = input("Seleccione una opción (1-3): ")

        if opcion == '1':
            convertirCF()
        elif opcion == '2':
            convertirFC()
        elif opcion == '3':
            print("Saliendo del programa... ¡Hasta luego!")
            print("Samuel Martinez NL = 35")
            break
        else:
            print("Opción no válida. Intente de nuevo.")

# Llamada principal para iniciar el programa
if __name__ == "__main__":
    menu()
````
